```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Resume Builder Heading
title = widgets.HTML(
    value="""
    <h1 style="color:#2563eb; text-align:center;">
        Resume Builder
    </h1>
    <p style="text-align:center;">
        Create Your Professional Resume Using Python
    </p>
    """
)

# Personal Information
name = widgets.Text(
    description="Full Name:",
    placeholder="Enter your full name"
)

email = widgets.Text(
    description="Email:",
    placeholder="Enter your email"
)

phone = widgets.Text(
    description="Phone:",
    placeholder="Enter your phone"
)

address = widgets.Text(
    description="Address:",
    placeholder="Enter your address"
)

# Education
education = widgets.Textarea(
    description="Education:",
    placeholder="Enter your education details",
    layout=widgets.Layout(width="500px", height="80px")
)

# Skills
skills = widgets.Textarea(
    description="Skills:",
    placeholder="Python, Java, SQL, HTML, CSS",
    layout=widgets.Layout(width="500px", height="80px")
)

# Projects
projects = widgets.Textarea(
    description="Projects:",
    placeholder="Enter your project details",
    layout=widgets.Layout(width="500px", height="80px")
)

# Experience
experience = widgets.Textarea(
    description="Experience:",
    placeholder="Enter internship or work experience",
    layout=widgets.Layout(width="500px", height="80px")
)

# Generate Button
generate_button = widgets.Button(
    description="Generate Resume",
    button_style="primary",
    icon="file"
)

output = widgets.Output()

def generate_resume(b):
    with output:
        clear_output()

        print("=" * 50)
        print("                 RESUME")
        print("=" * 50)

        print("\nPERSONAL INFORMATION")
        print("Name:", name.value)
        print("Email:", email.value)
        print("Phone:", phone.value)
        print("Address:", address.value)

        print("\nEDUCATION")
        print(education.value)

        print("\nSKILLS")
        print(skills.value)

        print("\nPROJECTS")
        print(projects.value)

        print("\nEXPERIENCE")
        print(experience.value)

        print("\n" + "=" * 50)

generate_button.on_click(generate_resume)

# Display Interface
display(
    title,
    name,
    email,
    phone,
    address,
    education,
    skills,
    projects,
    experience,
    generate_button,
    output
)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from reportlab.lib.pagesizes import A4
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, HRFlowable
)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import inch
from xml.sax.saxutils import escape
import os

# ---------- PROJECT TITLE ----------

title = widgets.HTML("""
<div style="background:#111827;padding:20px;border-radius:12px;">
    <h1 style="color:#60a5fa;text-align:center;">
        RESUME BUILDER
    </h1>
    <p style="color:white;text-align:center;">
        Create Your Professional Resume
    </p>
</div>
""")

# ---------- INPUT FIELDS ----------

def make_text(description, placeholder=""):
    return widgets.Text(
        description=description,
        placeholder=placeholder,
        layout=widgets.Layout(width="95%"),
        style={"description_width": "130px"}
    )

def make_area(description, placeholder=""):
    return widgets.Textarea(
        description=description,
        placeholder=placeholder,
        layout=widgets.Layout(width="95%", height="85px"),
        style={"description_width": "130px"}
    )

name = make_text("Full Name:", "Enter your full name")
email = make_text("Email:", "Enter your email")
phone = make_text("Phone:", "Enter your phone number")
address = make_text("Address:", "Enter your city")

education = make_area(
    "Education:",
    "MCA, University name, Passing year"
)

skills = make_area(
    "Skills:",
    "Python, SQL, HTML, CSS, JavaScript"
)

experience = make_area(
    "Experience:",
    "Internship or work experience"
)

projects = make_area(
    "Projects:",
    "Project name and description"
)

certifications = make_area(
    "Certifications:",
    "Course or certification name"
)

summary = make_area(
    "Profile Summary:",
    "Write a short professional summary"
)

# ---------- PDF GENERATION ----------

generate_button = widgets.Button(
    description="Generate PDF Resume",
    button_style="primary",
    icon="check",
    layout=widgets.Layout(width="250px", height="45px")
)

output = widgets.Output()

def safe_text(value):
    """Escape text so special characters don't break PDF paragraphs."""
    return escape(value.strip()).replace("\n", "<br/>") or "Not provided"

def generate_resume(button):
    with output:
        clear_output(wait=True)

        if not name.value.strip():
            print("Please enter your full name first.")
            return

        filename = "Professional_Resume.pdf"
        doc = SimpleDocTemplate(
            filename,
            pagesize=A4,
            rightMargin=0.65 * inch,
            leftMargin=0.65 * inch,
            topMargin=0.55 * inch,
            bottomMargin=0.55 * inch
        )

        styles = getSampleStyleSheet()

        styles.add(ParagraphStyle(
            name="ResumeName",
            parent=styles["Title"],
            fontSize=23,
            leading=28,
            alignment=TA_CENTER,
            textColor=colors.HexColor("#2563eb"),
            spaceAfter=8
        ))

        styles.add(ParagraphStyle(
            name="SectionHeading",
            parent=styles["Heading2"],
            fontSize=13,
            leading=17,
            textColor=colors.HexColor("#2563eb"),
            spaceBefore=12,
            spaceAfter=5
        ))

        styles.add(ParagraphStyle(
            name="ResumeBody",
            parent=styles["BodyText"],
            fontSize=10,
            leading=15,
            textColor=colors.HexColor("#222222"),
            spaceAfter=4
        ))

        story = []

        # Name and contact details
        story.append(Paragraph(safe_text(name.value), styles["ResumeName"]))

        contact = " | ".join(
            escape(value.strip())
            for value in [email.value, phone.value, address.value]
            if value.strip()
        )

        if contact:
            story.append(Paragraph(
                contact,
                ParagraphStyle(
                    "Contact",
                    parent=styles["ResumeBody"],
                    alignment=TA_CENTER
                )
            ))

        story.append(Spacer(1, 10))
        story.append(HRFlowable(
            width="100%",
            thickness=2,
            color=colors.HexColor("#2563eb")
        ))

        # Profile summary
        sections = [
            ("PROFESSIONAL SUMMARY", summary.value),
            ("EDUCATION", education.value),
            ("TECHNICAL SKILLS", skills.value),
            ("INTERNSHIP / EXPERIENCE", experience.value),
            ("PROJECTS", projects.value),
            ("CERTIFICATIONS", certifications.value)
        ]

        for heading, content in sections:
            if content.strip():
                story.append(Paragraph(heading, styles["SectionHeading"]))
                story.append(Paragraph(
                    safe_text(content),
                    styles["ResumeBody"]
                ))

        try:
            doc.build(story)

            print("Resume generated successfully!")
            print("File location:", os.path.abspath(filename))
            print("Open the file from your Jupyter Notebook folder.")

        except Exception as error:
            print("PDF generation error:", error)

generate_button.on_click(generate_resume)

# ---------- DISPLAY FORM ----------

display(title)

display(widgets.HTML("<h3>Personal Information</h3>"))
display(name, email, phone, address)

display(widgets.HTML("<h3>Professional Details</h3>"))
display(summary, education, skills, experience, projects, certifications)

display(generate_button, output)
```
